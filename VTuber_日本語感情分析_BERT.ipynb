```python
texts = [
    "今日の配信最高だった！",
    "次の配信は何時ですか？",
    "今日はちょっと残念だった"
]
```

```python

from bert_sentiment_analysis import BERTSentimentAnalyzer

# 加载日语情感分析模型
analyzer = BERTSentimentAnalyzer()

# 分析刚才的三条评论
results = analyzer.predict_batch(texts)

# 显示结果
for result in results:
    analyzer.display_result(result)
```

```python

test_texts = [
    "草ｗｗｗｗｗ",
    "え、神すぎるんだけど",
    "はいはい、今日も素晴らしい配信ですね（笑）",
    "5000円投げちゃった！",
    "かわいいいいいい！！！😭😭😭"
]

results = analyzer.predict_batch(test_texts)

for result in results:
    analyzer.display_result(result)
```
